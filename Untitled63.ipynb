```python

# ================================================================
# LAB / EXPERIMENT 17
# TEXT CLASSIFICATION USING A PRE-TRAINED TRANSFORMER (RoBERTa)
# ================================================================

# STEP 1: INSTALL REQUIRED LIBRARIES

!pip install -q transformers scikit-learn pandas matplotlib seaborn tqdm


# STEP 2: IMPORT LIBRARIES

import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn

from torch.utils.data import Dataset, DataLoader

from transformers import (
    AutoTokenizer,
    AutoModel,
    get_linear_schedule_with_warmup
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

from tqdm.auto import tqdm


# STEP 3: CONFIGURATION

TRAIN_PATH = "/content/training.csv"
VAL_PATH = "/content/validation.csv"
TEST_PATH = "/content/test.csv"

MODEL_NAME = "roberta-base"

MAX_LENGTH = 128
BATCH_SIZE = 16
EPOCHS = 15
LEARNING_RATE = 2e-5
DROPOUT = 0.3

OUTPUT_DIR = "/content/outputs"

os.makedirs(OUTPUT_DIR, exist_ok=True)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using device:", DEVICE)
print("Training epochs:", EPOCHS)


# STEP 4: SET RANDOM SEED

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


# STEP 5: LOAD CSV DATASETS

def load_csv(path):

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Dataset file not found: {path}"
        )

    df = pd.read_csv(path)

    print("\nLoading:", path)
    print("Columns:", list(df.columns))
    print("Shape:", df.shape)

    # Detect text column
    text_candidates = [
        "text", "sentence", "content",
        "tweet", "review", "utterance"
    ]

    # Detect label column
    label_candidates = [
        "label", "emotion", "sentiment",
        "target", "class"
    ]

    lower_columns = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    text_col = None
    label_col = None

    for col in text_candidates:
        if col in lower_columns:
            text_col = lower_columns[col]
            break

    for col in label_candidates:
        if col in lower_columns:
            label_col = lower_columns[col]
            break

    # Fallback if column names are different
    available = [
        c for c in df.columns
        if not str(c).lower().startswith("unnamed")
    ]

    if text_col is None and len(available) >= 1:
        text_col = available[0]

    if label_col is None:
        label_col = next(
            (c for c in available if c != text_col),
            None
        )

    if text_col is None or label_col is None:
        raise ValueError(
            "Could not identify text and label columns."
        )

    if text_col == label_col:
        raise ValueError(
            "Text and label columns cannot be the same."
        )

    df = df[[text_col, label_col]].copy()

    df.columns = ["text", "label"]

    # Remove missing values
    df = df.dropna()

    df["text"] = df["text"].astype(str).str.strip()

    # Normalize labels
    def normalize_label(value):

        if isinstance(value, (int, np.integer)):
            return str(int(value))

        if isinstance(value, (float, np.floating)):
            if float(value).is_integer():
                return str(int(value))

        return str(value).strip()

    df["label"] = df["label"].apply(normalize_label)

    df = df[
        (df["text"] != "") &
        (df["label"] != "")
    ]

    return df.reset_index(drop=True)


train_df = load_csv(TRAIN_PATH)
val_df = load_csv(VAL_PATH)
test_df = load_csv(TEST_PATH)


# STEP 6: DATASET INFORMATION

print("\n========== DATASET INFORMATION ==========")

print("Training samples:", len(train_df))
print("Validation samples:", len(val_df))
print("Testing samples:", len(test_df))

print("\nTraining data:")
print(train_df.head())

print("\nEmotion distribution:")
print(train_df["label"].value_counts())


# STEP 7: LABEL ENCODING

all_labels = set(train_df["label"])

def label_sort_key(label):
    try:
        return (0, float(label))
    except ValueError:
        return (1, label.lower())

classes = sorted(all_labels, key=label_sort_key)

label2id = {
    label: index
    for index, label in enumerate(classes)
}

id2label = {
    index: label
    for label, index in label2id.items()
}

NUM_CLASSES = len(classes)

for df in [train_df, val_df, test_df]:

    unknown_labels = set(df["label"]) - set(label2id)

    if unknown_labels:
        raise ValueError(
            f"Unknown labels found: {unknown_labels}"
        )

    df["label_id"] = df["label"].map(label2id).astype(int)

print("\nNumber of classes:", NUM_CLASSES)
print("Label mapping:", label2id)


# STEP 8: LOAD PRE-TRAINED RoBERTa TOKENIZER

print("\nLoading RoBERTa tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)


# STEP 9: CREATE PYTORCH DATASET

class EmotionDataset(Dataset):

    def __init__(self, texts, labels):

        self.texts = list(texts)
        self.labels = list(labels)

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, index):

        text = str(self.texts[index])

        encoding = tokenizer(
            text,
            max_length=MAX_LENGTH,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        return {
            "input_ids": encoding["input_ids"].squeeze(0),

            "attention_mask": encoding[
                "attention_mask"
            ].squeeze(0),

            "labels": torch.tensor(
                int(self.labels[index]),
                dtype=torch.long
            )
        }


train_dataset = EmotionDataset(
    train_df["text"],
    train_df["label_id"]
)

val_dataset = EmotionDataset(
    val_df["text"],
    val_df["label_id"]
)

test_dataset = EmotionDataset(
    test_df["text"],
    test_df["label_id"]
)


# STEP 10: CREATE DATA LOADERS

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)


# STEP 11: IMPLEMENT RoBERTa PYTORCH ARCHITECTURE

class RobertaClassifier(nn.Module):

    def __init__(self, num_classes):

        super(RobertaClassifier, self).__init__()

        self.roberta = AutoModel.from_pretrained(
            MODEL_NAME
        )

        hidden_size = self.roberta.config.hidden_size

        self.dropout = nn.Dropout(DROPOUT)

        self.classifier = nn.Linear(
            hidden_size,
            num_classes
        )

    def forward(self, input_ids, attention_mask):

        outputs = self.roberta(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # Extract first token representation
        pooled_output = outputs.last_hidden_state[:, 0, :]

        pooled_output = self.dropout(pooled_output)

        logits = self.classifier(pooled_output)

        return logits


model = RobertaClassifier(NUM_CLASSES)

model = model.to(DEVICE)

print("\nRoBERTa model initialized successfully!")


# STEP 12: LOSS FUNCTION AND OPTIMIZER

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=0.01
)

total_steps = len(train_loader) * EPOCHS

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_steps),
    num_training_steps=total_steps
)


# STEP 13: TRAINING AND EVALUATION FUNCTION

def run_epoch(loader, training=False):

    if training:
        model.train()
    else:
        model.eval()

    total_loss = 0

    all_predictions = []
    all_labels = []

    context = (
        torch.enable_grad()
        if training
        else torch.no_grad()
    )

    with context:

        for batch in tqdm(loader):

            input_ids = batch["input_ids"].to(DEVICE)

            attention_mask = batch[
                "attention_mask"
            ].to(DEVICE)

            labels = batch["labels"].to(DEVICE)

            if training:
                optimizer.zero_grad()

            outputs = model(
                input_ids=input_ids,
                attention_mask=attention_mask
            )

            loss = criterion(outputs, labels)

            if training:

                loss.backward()

                torch.nn.utils.clip_grad_norm_(
                    model.parameters(),
                    max_norm=1.0
                )

                optimizer.step()
                scheduler.step()

            total_loss += loss.item()

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            all_predictions.extend(
                predictions.detach().cpu().numpy()
            )

            all_labels.extend(
                labels.detach().cpu().numpy()
            )

    average_loss = total_loss / max(len(loader), 1)

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    f1 = f1_score(
        all_labels,
        all_predictions,
        average="weighted",
        zero_division=0
    )

    return (
        average_loss,
        accuracy,
        f1,
        all_labels,
        all_predictions
    )


# STEP 14: TRAIN RoBERTa FOR 15 EPOCHS

history = {
    "train_loss": [],
    "val_loss": [],
    "train_accuracy": [],
    "val_accuracy": [],
    "train_f1": [],
    "val_f1": []
}

best_val_f1 = -1

best_model_path = os.path.join(
    OUTPUT_DIR,
    "best_roberta_model.pt"
)

print("\n========== STARTING 15 EPOCH TRAINING ==========")

for epoch in range(EPOCHS):

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")

    train_loss, train_acc, train_f1, _, _ = run_epoch(
        train_loader,
        training=True
    )

    val_loss, val_acc, val_f1, _, _ = run_epoch(
        val_loader,
        training=False
    )

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)

    history["train_accuracy"].append(train_acc)
    history["val_accuracy"].append(val_acc)

    history["train_f1"].append(train_f1)
    history["val_f1"].append(val_f1)

    print("\nTraining Loss:", round(train_loss, 4))
    print("Training Accuracy:", round(train_acc, 4))
    print("Training F1:", round(train_f1, 4))

    print("\nValidation Loss:", round(val_loss, 4))
    print("Validation Accuracy:", round(val_acc, 4))
    print("Validation F1:", round(val_f1, 4))

    # Save best model based on validation F1
    if val_f1 > best_val_f1:

        best_val_f1 = val_f1

        torch.save(
            model.state_dict(),
            best_model_path
        )

        print("Best model saved!")


# STEP 15: LOAD BEST MODEL

model.load_state_dict(
    torch.load(
        best_model_path,
        map_location=DEVICE
    )
)

print("\nBest RoBERTa model loaded!")


# STEP 16: TEST EVALUATION

print("\n========== RoBERTa TEST EVALUATION ==========")

test_loss, test_accuracy, test_f1, y_true, y_pred = run_epoch(
    test_loader,
    training=False
)

test_precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

test_recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\nTest Loss:", round(test_loss, 4))
print("Accuracy:", round(test_accuracy, 4))
print("Precision:", round(test_precision, 4))
print("Recall:", round(test_recall, 4))
print("F1 Score:", round(test_f1, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        target_names=classes,
        zero_division=0
    )
)


# STEP 17: TRAINING AND VALIDATION LOSS GRAPH

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, EPOCHS + 1),
    history["train_loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    range(1, EPOCHS + 1),
    history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title("RoBERTa Training and Validation Loss")

plt.legend()
plt.grid(alpha=0.3)

plt.xticks(range(1, EPOCHS + 1))

plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "loss_curves.png")
)

plt.show()


# STEP 18: TRAINING AND VALIDATION ACCURACY GRAPH

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, EPOCHS + 1),
    history["train_accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    range(1, EPOCHS + 1),
    history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title("RoBERTa Training and Validation Accuracy")

plt.legend()
plt.grid(alpha=0.3)

plt.xticks(range(1, EPOCHS + 1))

plt.tight_layout()

plt.savefig(
    os.path.join(OUTPUT_DIR, "accuracy_graph.png")
)

plt.show()


# STEP 19: RoBERTa CONFUSION MATRIX

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES))
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=classes,
    yticklabels=classes
)

plt.xlabel("Predicted Emotion")
plt.ylabel("Actual Emotion")

plt.title("RoBERTa Confusion Matrix")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "roberta_confusion_matrix.png"
    )
)

plt.show()


# STEP 20: BASELINE MODEL - TF-IDF + LOGISTIC REGRESSION

print("\n========== TRAINING BASELINE MODEL ==========")

baseline_model = Pipeline([

    (
        "tfidf",
        TfidfVectorizer(
            max_features=30000,
            ngram_range=(1, 2),
            sublinear_tf=True
        )
    ),

    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=SEED
        )
    )

])

baseline_model.fit(
    train_df["text"],
    train_df["label_id"]
)

baseline_predictions = baseline_model.predict(
    test_df["text"]
)


# STEP 21: BASELINE MODEL EVALUATION

baseline_accuracy = accuracy_score(
    test_df["label_id"],
    baseline_predictions
)

baseline_precision = precision_score(
    test_df["label_id"],
    baseline_predictions,
    average="weighted",
    zero_division=0
)

baseline_recall = recall_score(
    test_df["label_id"],
    baseline_predictions,
    average="weighted",
    zero_division=0
)

baseline_f1 = f1_score(
    test_df["label_id"],
    baseline_predictions,
    average="weighted",
    zero_division=0
)

print("\n========== BASELINE RESULTS ==========")

print("Accuracy:", round(baseline_accuracy, 4))
print("Precision:", round(baseline_precision, 4))
print("Recall:", round(baseline_recall, 4))
print("F1 Score:", round(baseline_f1, 4))


# STEP 22: BASELINE CONFUSION MATRIX

baseline_cm = confusion_matrix(
    test_df["label_id"],
    baseline_predictions,
    labels=list(range(NUM_CLASSES))
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    baseline_cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=classes,
    yticklabels=classes
)

plt.xlabel("Predicted Emotion")
plt.ylabel("Actual Emotion")

plt.title("Baseline Confusion Matrix")

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "baseline_confusion_matrix.png"
    )
)

plt.show()


# STEP 23: COMPARE RoBERTa AND BASELINE

comparison = pd.DataFrame({

    "Model": [
        "TF-IDF + Logistic Regression",
        "RoBERTa"
    ],

    "Accuracy": [
        baseline_accuracy,
        test_accuracy
    ],

    "Precision": [
        baseline_precision,
        test_precision
    ],

    "Recall": [
        baseline_recall,
        test_recall
    ],

    "F1 Score": [
        baseline_f1,
        test_f1
    ]

})

print("\n========== MODEL COMPARISON ==========")

print(comparison.to_string(index=False))

comparison.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "model_comparison.csv"
    ),
    index=False
)


# STEP 24: MODEL COMPARISON GRAPH

comparison_plot = comparison.set_index("Model")

comparison_plot.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Baseline vs RoBERTa Performance")

plt.ylabel("Performance Score")

plt.ylim(0, 1)

plt.xticks(rotation=0)

plt.legend(loc="lower right")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "model_comparison.png"
    )
)

plt.show()


# STEP 25: DISPLAY SAMPLE PREDICTIONS

print("\n========== SAMPLE PREDICTIONS ==========")

for i in range(min(10, len(y_true))):

    actual = id2label[int(y_true[i])]

    predicted = id2label[int(y_pred[i])]

    print("\nText:", test_df.iloc[i]["text"])

    print("Actual Emotion:", actual)

    print("Predicted Emotion:", predicted)


# STEP 26: SAVE PREDICTIONS

prediction_df = test_df[["text", "label"]].copy()

prediction_df["Actual Emotion"] = [
    id2label[int(x)] for x in y_true
]

prediction_df["Predicted Emotion"] = [
    id2label[int(x)] for x in y_pred
]

prediction_df["Correct"] = (
    prediction_df["Actual Emotion"] ==
    prediction_df["Predicted Emotion"]
)

prediction_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "test_predictions.csv"
    ),
    index=False
)


# STEP 27: CUSTOM TEXT PREDICTION

def predict_emotion(text):

    model.eval()

    encoding = tokenizer(
        text,
        max_length=MAX_LENGTH,
        padding="max_length",
        truncation=True,
        return_tensors="pt"
    )

    input_ids = encoding["input_ids"].to(DEVICE)

    attention_mask = encoding[
        "attention_mask"
    ].to(DEVICE)

    with torch.no_grad():

        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        prediction = torch.argmax(
            probabilities,
            dim=1
        ).item()

    emotion = id2label[prediction]

    confidence = probabilities[0][prediction].item()

    return emotion, confidence


print("\n========== CUSTOM TEXT PREDICTION ==========")

sample_text = "I am feeling very happy today!"

emotion, confidence = predict_emotion(sample_text)

print("Input:", sample_text)

print("Predicted Emotion:", emotion)

print("Confidence:", round(confidence * 100, 2), "%")


# STEP 28: SAVE MODEL AND TOKENIZER

torch.save(
    model.state_dict(),
    os.path.join(
        OUTPUT_DIR,
        "final_roberta_model.pt"
    )
)

tokenizer.save_pretrained(
    os.path.join(
        OUTPUT_DIR,
        "tokenizer"
    )
)

pd.DataFrame(history).to_csv(
    os.path.join(
        OUTPUT_DIR,
        "training_history.csv"
    ),
    index=False
)


# STEP 29: FINAL RESULT

print("\n========================================")

print("EXPERIMENT 17 COMPLETED SUCCESSFULLY!")

print("========================================")

print("\nTotal epochs:", EPOCHS)

print("RoBERTa Test Accuracy:", round(test_accuracy, 4))

print("Baseline Test Accuracy:", round(baseline_accuracy, 4))

if test_accuracy > baseline_accuracy:

    print("\nRoBERTa performed better than the baseline model.")

else:

    print("\nBaseline performed equally or better than RoBERTa.")

print("\nGenerated output files:")

for file in sorted(os.listdir(OUTPUT_DIR)):
    print(file)

print("\nAll results saved in:", OUTPUT_DIR)
```
